```python
plt.figure(figsize = (8,5))
plt.hist(df["math score"], bins = 10, edgecolor = "black")
plt.title("Distribution of Math Score")
plt.xlabel("Math Marks")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python
for subject in ["math score", "reading score", "writing score"]:
    plt.figure(figsize = (7,4))
    plt.hist(df[subject], bins = 10 , edgecolor = "black")
    plt.title(f"Distribution of {subject.title()}")
    plt.xlabel("Marks")
    plt.ylabel("Number of Students")
    plt.show()
```

```python
averages = df[["math score", "reading score", "writing score"]].mean()
averages.plot(kind="bar", figsize=(7,4), edgecolor="black")

plt.title("Average Score by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.xticks(rotation = 0)
plt.show()
```

```python
gender_avg = df.groupby("gender")[
    ["math score", "reading score", "writing score"]
 ].mean()

gender_avg.plot(kind = "bar", figsize=(8,5))
plt.title("Average Scores By Gender")
plt.xlabel("Gender")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.legend(title="Subject")
plt.tight_layout()
plt.show()
```

```python
prep_avg = df.groupby("test preparation course")[
    ["math score", "reading score", "writing score"]
 ].mean()

prep_avg.plot(kind = "bar", figsize=(8,5))
plt.title("Average Scores By Test Prepration Course")
plt.xlabel("Test Preparation Course")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.legend(title="Subject")
plt.tight_layout()
plt.show()
```

```python
df["parental level of education"].value_counts()
```

```python
parent_avg = df.groupby("parental level of education") [
    ["math score", "reading score", "writing score"]
].mean()

parent_avg
```

```python
parent_avg.plot(kind = "bar", figsize=(11,6))

plt.title("Average Student Scores by Parental Education")
plt.xlabel("Parental Level of Education")
plt.ylabel("Average Marks")
plt.xticks(rotation = 35, ha="right")
plt.legend(title="Subject")
plt.tight_layout()
plt.show()
```

```python
df["lunch"].value_counts()
```

```python
lunch_avg = df.groupby("lunch")[
    ["math score", "reading score", "writing score"]
].mean()
lunch_avg
```

```python
lunch_avg.plot(kind = "bar", figsize=(8,5))
plt.title("Average Student Scores by Lunch Category")
plt.xlabel("Lunch Category")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
lunch_difference = lunch_avg.loc["standard"] - lunch_avg.loc["free/reduced"]
print("Score difference (standard - free/reduced) :")
print(lunch_difference)
```
